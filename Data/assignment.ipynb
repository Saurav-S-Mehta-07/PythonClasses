## cleaning and structuring our data

```python
with open("store_data.json",'r') as f:
    data = json.load(f)
```

```python

def clean_data():
    
    text_to_num = {"one":1, "two":2,"three":3,"four":4, "five":5}

    cleaned_data = []
    unique_user = set()


    for d in data:
    
        #name space removed
        d['name'] = d['name'].strip()

        # rating to float values 
        if(text_to_num.get(d['rating'])):
           d['rating'] = float(text_to_num[d['rating']])
        else:
           d['rating'] = float(d['rating'])

        # age to numbers
        if(d.get('age')):
          d['age'] = int(d['age'])
        else:
            d['age'] = None

        
        if(d['name'] not in unique_user):
           cleaned_data.append(d)
            
        unique_user.add(d['name'])
        
        
    return cleaned_data
```

```python
data = clean_data()
for d in data:
    print(d, end = '\n\n')
```

# \#<u><b> <i> Assignment 2 </i></b></u>
## getting meaningful insights from data
1. What's avg rating?
2. What % of users gave poor rating (<3)?

```python
def get_insights(data):
    total_rating = 0
    for user in data:
        total_rating += user['rating']
    avg_rating = total_rating / len(data)
    print(f"avg rating is {avg_rating}")

    poor_rating = 0
    for user in data:
        if(user['rating']<3):
            poor_rating += 1

    poor_rating_percent = poor_rating/len(data) * 100
    print(f"{poor_rating_percent} % of the  users gave poor rating")

get_insights(data)
```

# \#<u><b> <i> Assignment 3 </i></b></u>
## build a product recommendation feature (rule based, not ML)
1. If user rating >= 4, recommend some brand products
2. If user rating < 4, recommend different brand products.

```python
def get_recommendation(data):
    recommendations = []

    for user in data:
        curr_recommendation = {}
        curr_recommendation['name'] = user['name']
        
        if(user['rating'] >= 4):
            curr_recommendation['Brand'] = "Apple"
        else:
            curr_recommendation['Brand'] = "SamSung"
            
        recommendations.append(curr_recommendation)
        
    return recommendations

get_recommendation(data)
```
