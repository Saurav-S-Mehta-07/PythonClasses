# \# File I/O

```python
# read from file
f = open("sample.txt","r")
data = f.read()
print(data)
f.close()
```

```python
# with open
with open("sample.txt","r") as f:
    data = f.readline() # read one line
    print(data, end='')
```

```python
# write
with open("sample.txt","w") as f:
    f.write("Text using write w mode \n this is 2nd line")

with open("sample.txt", "r") as f : 
    data = f.read()
    print(data)
```

```python
with open("sample2.txt","x") as f:
    f.write("new file created")
```

# search a word in sample.txt

```python
word = input("enter word : ")
with open("sample.txt","r") as f :
    line_count = 1
    flag = 0
    for line in f.readlines():
        if(word in line):
            print(f"{word} is present in line {line_count}")
            flag = 1
        line_count+=1
    if(flag == 0):
        print("word not found in the file")
```

# \# Exception Handling

```python
try:
    a = 100
    b = 0
    res = a/b
except:
    print("can't divide by 0")
else:
    print(f"on dividing {a} by {b} result is: {res}")
finally:
    print("timeout now")
```

# \# List Comprehensions

```python
# square for 1 to 5 in a list
list = [i * i  for i in range(1,6) ]
list

# multiple with 2 
list = [i*2 for i in range(1,6)]
list

# all even form 1 to 10
list = [ i for i in range(1,11) if i%2==0]
list
```

```python
# change --ve no. to 0
list = [-2, -4, 3, 5, 2, -1]

list = [0 if (i<0) else i for i in list]
list
```

```python
# word in upper case
words = ["hello", "python", "world"]
words = [word.upper() for word in words]
words
```

# \# JSON Module

```python
import json
```

```python
data = {
    "name" : "Saurav",
    "age" : 19,
    "isStudent" : True
}

with open("data.json", "w") as f : 
    json.dump(data,f, indent=4, sort_keys=True)

with open("data.json","r") as f : 
    py_obj = json.load(f)
    print(py_obj)
```
