```python
#find average of 3 subjects marks
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def average(self):
        sum = 0
        for mark in self.marks:
            sum+=mark
        avg = sum/len(self.marks)

        print(f"hey {self.name}, your marks's avg is {avg}")

s1 = Student("Saurav",[99,97,98])
s2 = Student("Mayank", [100,33,88])

s1.average()
s2.average()
```

```python
#methods

class Student:
    college = "GEHU"
    
    def __init__(self, name, course):
        self.name = name
        self.course = course
        
    #class method
    @classmethod
    def get_college_name(cls):
        print(f"college name is : {cls.college}")

    #instance method
    def info(self):
        print(f"name : {self.name}, course : {self.course}, college : {self.college}")

    #static method
    @staticmethod
    def greet():
        print(f"Hello! Student, Welcome GEHU")
        
s1 = Student("Saurav", "BCA")
s1.info()

Student.get_college_name()

s1.greet()
```

# problem solution - product store
* design and create an online store for products (name, price).
* track total products being created
* create a static method to calculate discount on each product based on a % parameter.

```python
class Product:
    count_product = 0
    def __init__(self, name, price):
        self.name = name
        self.price = price
        Product.count_product+=1
    @staticmethod
    def discount(name,price):
        if(price>50000):
            dis = 20/100 * price
            print(f"This product ({name}) have discount of Rs. {dis}")
        else:
            dis = 5/100 * price
            print(f"This product ({name}) have discount of Rs. {dis}")
            
            

p1 = Product("Laptop",70000)
p2 = Product("Tablet",50000)
p3 = Product("Mobile",10000)

Product.count_product

p1.discount(p1.name,p1.price)
p2.discount(p2.name,p2.price)
p3.discount(p3.name,p3.price)

```

# --->>> Encapsulation

```python
class User:
    
    def __init__(self, name,balance, password):
        self.name = name # public attribute
        self.__password = password # private attr.
        self._balance = balance # protected attr.
        
    def get_password(self): # data mangling
        return self.__password

u1 = User("Saurav",20000, 1234)
u1.name
u1._balance

u1.get_password() # data mangling
```

# --->>> Inheritance

```python
class Father:
    eye_color = "blue"
    
class Mother:
    hair_color = "brown"
    eye_color = "green"
    
class Child(Father, Mother):
    def get_eye_color(self):
        print(self.eye_color)
    def get_hair_color(self):
        print(self.hair_color)

c = Child()
c.get_eye_color()
c.get_hair_color()
```

# --->>> Abstraction

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Lion(Animal):
    def make_sound(self):
        print("Lion sound")

l = Lion()
l.make_sound()
```

# --->>> Polymorphism

```python
# function overriding
class Parent:
    def eye_color(self):
        print("blue")
class Child(Parent):
    def eye_color(self):
        print("green")

c = Child()
c.eye_color()
```

```python
# duck typing
class Teacher:
    def get_designation(self):
        print("designation : Teacher")

class Accountant:
    def get_designation(self):
        print("designation : Accountant")

t1 = Teacher()
t1.get_designation()

a1 = Accountant()
a1.get_designation()
```
