# problem solution

```python
info = [
    ("Alice", "Math"),
    ("Bob", "Science"),
    ("Charlie", "Math"),
    ("Alice", "Science"),
    ("Bob", "Math"),
    ("Alice", "English"),
    ("Charlie", "English")
]
```

```python
# list all unique course
course = set()
for val in info:
     course.add(val[1])
course
```

```python
# lsit students enrolled in English
for std in info:
    if std[1]=="English":
        print(std[0])
```

```python
# create a dictionary (student, set of courses)
dict = {}
for std in info:
    dict[std[0]] = set()
dict
for std in info:
    dict[std[0]].add(std[1])

dict
```
