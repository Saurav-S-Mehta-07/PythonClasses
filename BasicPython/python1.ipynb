```python
# arithmetic
a, b = 10, 5
print(a+b, a-b, a*b, a/b, a%b, a**b)

#relational/comparison
print(a>b, b>a, b==a, b>=a, b<=a)

#assignment operators
c = b
a += c
print(a)

#logical operator
c = not a
print(c)

b = 0
a = 3
print(a and b)
print(a or b)
```
