# Control statements

## 1. Conditional Statements

```python
age = int(input("enter your age : "))

if(age>=100):
    print("wrong age")
elif(age>=18):
    print("adult")
elif(age>=13):
    print("teen")
else:
    print("child")
```

## 2. loop statements

```python
for i in range(0,11,1):
    print(i, end=" ")

print("\n")

a = 5
while a!=0:
    print(a, end=' | ')
    a -= 1
```

```python
# match case
n  = int(input ("enter day number : "))

match n:
    case 1 : 
        print("Monday")
    case 2 :
        print("Tuesday")
    case 3 : 
        print("Wednesday")
    case 4 : 
        print("Thursday")
    case 5 : 
        print("Friday")
    case 6 :
        print("Saturday")
    case 7 :
        print("Sunday")
    case _:
        print("invalid day number")
```

```python
# print vowel count of a given count
word = "artificial"

count = 0
vowel = "aeiou"
for ch  in word :
    if ch in vowel:
        count+=1
print(f"total vowel in {word} : {count}")
```

## 3. jump statement

```python
for i in range(1,10):
    if i==5:
        break
    if(i==2):
        continue
    print(i, end = " ")
```

## Functions

```python
def hello():
    print("Hello World!!!")
hello()
hello()
```

```python
def sum(a,b):
    return a+b
print(sum(4,5))
```

```python
def product(a, b =1):
    return a*b
print(product(3,5))
print(product(5))
```

```python
# labda function

res = lambda a,b,c : a+b+c
print(res(4,5,6))
```

```python
# factorial number

def fact(n):
    if(n==0):
        return 1
    return n * fact(n-1)
print(fact(5))
```
