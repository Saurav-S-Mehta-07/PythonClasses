# \# Binomial Distribution 

```python
from scipy.stats import binom
```

```python
# a fair coin is tossed 5 times find the probability that excatly 3 heads are obtained
n = 5
k = 3
p = 0.5

prob = binom.pmf(k,n,p)
print(prob)
```

# \# Unifom distribution

```python
import numpy as np
import matplotlib.pyplot as plt
values = np.random.uniform(0,10,1000000)
plt.hist(values, bins=100, density=True, alpha=0.3)
plt.title("Continuous UD")
plt.xlabel("x")
plt.ylabel("Probabily density")
plt.grid(True)
plt.show()
```

# \# normal distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

u = 70 # mean
sigma = 10

x = np.linspace(30,110, 1000)
y = norm.pdf(x,u,sigma)
plt.plot(x,y,color="black", linewidth=2, label="Noraml Distribution")

x3 = np.linspace(u-3*sigma,u+3*sigma,1000)
plt.fill_between(x3, norm.pdf(x3,u,sigma),color="red", alpha=0.3, label="97.8%")

x2 = np.linspace(u-2*sigma,u+2*sigma,1000)
plt.fill_between(x2, norm.pdf(x2,u,sigma),color="blue", alpha=0.3, label="95%")

x1 = np.linspace(u-sigma,u+sigma,1000)
plt.fill_between(x1, norm.pdf(x1,u,sigma),color="green", alpha=0.3, label="68%")

plt.title("normal distribution")
plt.legend()
plt.show()
```
