# ploting y =  sqrt(x)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,100,100000)
y = np.sqrt(x)
plt.plot(x,y,label="y=sqrt(x)")
plt.show()
```

# sin x

```python
# create x values from -2pi to 2pi
x = np.linspace(-2*np.pi, 2*np.pi,100)
y = np.sin(x)
plt.plot(x,y,label="y = sin(x)", color = "blue")

plt.axvline(0,color="black", linewidth=1.5,linestyle='--', label = "x=0")
plt.axhline(0,color="black", linewidth=1.5, linestyle='--', label="y=0")

plt.text(0.1,-0.07,"0",ha='center')

plt.legend()
plt.show()
```

```python
# scaler mulitply
x = np.linspace(-2*np.pi, 2*np.pi,100)
y = 2 * np.sin(x)
plt.plot(x,y,label="y = 2 * sin(x)", color = "blue")

plt.axvline(0,color="black", linewidth=1.5,linestyle='--', label = "x=0")
plt.axhline(0,color="black", linewidth=1.5, linestyle='--', label="y=0")

plt.text(0.1,-0.07,"0",ha='center')

plt.legend()
plt.show()
```

```python
# 1/2
x = np.linspace(-2*np.pi, 2*np.pi,100)
y = np.sin(x) / 2
plt.plot(x,y,label="y = 1/2  * sin(x)", color = "blue")

plt.axvline(0,color="black", linewidth=1.5,linestyle='--', label = "x=0")
plt.axhline(0,color="black", linewidth=1.5, linestyle='--', label="y=0")

plt.text(0.1,-0.07,"0",ha='center')

plt.legend()
plt.show()
```

```python
# scalar addition
x = np.linspace(-2*np.pi, 2*np.pi,100)

y = np.sin(x) + 2

plt.plot(x,y,label="y = sin(x) + 2", color = "blue")

plt.axvline(0,color="black", linewidth=1.5,linestyle='--', label = "x=0")
plt.axhline(0,color="black", linewidth=1.5, linestyle='--', label="y=0")

plt.text(0.1,-0.07,"0",ha='center')

plt.legend()
plt.show()
```

```python
# scalar subtraction
x = np.linspace(-2*np.pi, 2*np.pi,100)

y = np.sin(x) - 2


plt.plot(x,y,label="y = sin(x) - 2", color = "blue")

plt.axvline(0,color="black", linewidth=1.5,linestyle='--', label = "x=0")
plt.axhline(0,color="black", linewidth=1.5, linestyle='--', label="y=0")

plt.text(0.1,-0.07,"0",ha='center')

plt.legend()
plt.show()
```

```python
# f(x) = x^2

x = np.linspace(-100,100,200)
y = x ** 2

plt.plot(x,y, label="y = x^2")
plt.axvline(0, color="black", linestyle='--')
plt.legend()
plt.show()
```

```python
# f(x) = x^2
# x = x+5

x = np.linspace(-15,15,30)
y = (x+5)**2

plt.plot(x,y, label="y = (x+5)^2")
plt.axvline(0, color="black", linestyle='-')
plt.axvline(-5, color="black", linestyle='--')

plt.legend()
plt.show()
```
