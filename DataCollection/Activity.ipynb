## from quotes, extract all quotes & authors from all pages with the life tag

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# extracting html and saving
page_no = 1;
while True:
    url = f"https://quotes.toscrape.com/page/{page_no}/"
    res = requests.get(url)

    soup = BeautifulSoup(res.text,"lxml")
    quotes = soup.select("div .quote")
    if not quotes :
        print("not valid quotes anymore")
        break
    with open(f"scraped_data/data{page_no}","w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"downloaded data from {page_no}")
    page_no = page_no+1
```

```python
# extracting useful infomation
life_quotes = []

for i in range(1, page_no):
    with open(f"scraped_data/data{i}", "r", encoding="utf-8") as f :
        html_content = f.read()
    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")
    
    for q in all_quotes:
        all_tags  = []
        for tag in q.select("div.tags a.tag"):
            all_tags.append(tag.get_text())
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text,author])
```

```python
df = pd.DataFrame(life_quotes, columns=["quotes", "author"])
df
```

```python
df.to_csv("Cleaned_data/quotes.csv")
print("cleaned data saved")
```
