# \# <i> working with APIs

```python
import requests
url = "https://stephen-king-api.onrender.com/api/books"
res = requests.get(url)
res
```

```python
json_data = res.json()
# json_data
```

```python
# this json data to datagframe
import pandas as pd
df = pd.json_normalize(json_data["data"])
df.head()
```

```python
df = df[["id", "Title", "Year", "ISBN"]]
df.head()
```
