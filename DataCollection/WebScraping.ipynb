# \# <i> Web Scraping

```python
import requests
url = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(url)
res.status_code
```

```python
res.status_code
res.text # html in string
res.content # raw bytes of the response
res.headers # meta data like content-type date etc
```

# saving data

```python
with open("scraped_data/data.html", "w", encoding="utf-8") as f:
    f.write(res.text)
print("data saved")
```

```python
# try-except
try:
    r = requests.get(url, timeout=5)
    r.raise_for_status()
except requests.exceptions.RequestException as e :
    print("Error : " , e)
```

# \# working with beautifulSoup

```python
from bs4 import BeautifulSoup # class
```

```python
with open("scraped_data/data.html","r", encoding="utf-8") as f :
    html_content = f.read()

# creating an object of BeautifulSoup class
soup = BeautifulSoup(html_content, "lxml")
# soup
```

# \* BeautifulSoup's Methods

```python
soup.find("h1") # find first h1 tag
soup.h1 # same find first h1 tag

all_h3 = soup.find_all("h3") # find all h3 tag

for h3 in all_h3:
    print(h3.get_text(strip=True))
    break;

for h3  in all_h3:
    print(h3.find_parent("div")["class"])
    break;

```

```python
# css selector

for h3 in all_h3:
    print(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))
    print(h3.attrs)
    break;

# navigation dom
for h3 in all_h3:
    print("parent : \n " , h3.parent)
    print("children : \n", h3.children)
    break;
    
```

# making useful data now

```python
all_h3 = soup.find_all("h3")
type(all_h3)

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    all_countries.append([name,population])
```

```python
import pandas as pd
df = pd.DataFrame(all_countries, columns=["Name", "Population"])
df.to_csv("Cleaned_data/data.csv",index=False)
print("meaningful data is stored from scraping")
```
