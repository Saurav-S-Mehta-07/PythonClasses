# \# Activiy : AQI dataset

```python
import pandas as pd
```

```python
# aqi dataset
df = pd.read_csv("globalAirQuality.csv")
df.head()
df.tail()
```

```python
df.describe()
```

# selecting data

```python


# selecting columns
df["city"]
df[["city", "aqi"]]

# row
df.loc[1] # row 1
df.loc[0:2] # row 0 to 2 -> ending index is inclusive

df.loc[0:4:2]

df.iloc[0]
df.iloc[0:2] # ending index is exclusive
```

```python
# selecting row and col

df.loc[0, "aqi"] # 108
      #row, #col

df.loc[0, ["city", "aqi"]]

df.loc[0:2, ["city", "aqi"]]

df.iloc[0:3, [2,11]]
             #city, #aqi index
```

```python
# single scalar
df.at[0,"city"] # New York
df.iat[0,2] # New York
```

# filtering

```python
# country where aqi > 200
df[df["aqi"]>200] # return a view

# rows where aqi > 195 and temp >30
df[(df["aqi"]>195) & (df["temperature"]>30)]

# print only city and aqi of those who has aqi>100
df[df["aqi"]>100][["city", "aqi"]]
```

```python
# filtring using query
df.query("aqi>100 & temperature>30") # return a copy
```

```python
num = 30
df.query("aqi>200 and temperature>@num")
```

```python
df.query("~(aqi>200) and temperature > 30") # where there aqi not > 200 and temp >30
```
