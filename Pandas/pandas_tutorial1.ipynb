# \# Series

```python
import pandas as pd
s = pd.Series([1,2,3,4,5])
print(s)
print(type(s))
print(s[0])
```

```python
s = pd.Series([1,2,3,4,5], index=["adam", "bob", "charlie", "denis", "eve"])
print(s)
print(s["bob"])
print(s[1])
print(s.index)
```

```python
#vectorized operation
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([6,7,8,9,10])

print(s1 + s2)
```

```python
s = pd.Series([1,2,3,4,5])
s[0] = 100
print(s)

#drop
changed_s = s1.drop(0)
print(changed_s)
print(s)
```

# \# Dataframe

```python
# dataframe creation using dictionary
info = {
    "name" : ["adam", "bob", "eve"],
    "age" : [23,24,25],
    "gpa" : [9.5, 9.7,9.4]
}

df = pd.DataFrame(info)
print(df)
print(df.index) # row labels
print(df.columns) # col labels
```

```python
# datagram using list of list
df = pd.DataFrame([["Adam", 45], ["Bob", 23]], columns=["name", "age"])
df
```

```python
# dataframe using numpy array
import numpy as np

np_arr = np.array([[1,2,3],[4,5,6]])

df = pd.DataFrame(np_arr, columns=["a", "b", "c"])
df
```
