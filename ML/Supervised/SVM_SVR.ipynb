# \# SVR Implementation

```python
from sklearn.svm import SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn import datasets
```

```python
df = datasets.load_diabetes(as_frame=True).frame
df.head()
```

```python
X = df.drop("target", axis = 1)
y = df["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, random_state=42)
```

```python
y_scaler = StandardScaler()
y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled = y_scaler.transform(y_test.values.reshape(-1,1)).ravel()
```

```python
model = SVR() # default kernel = rbf
model.fit(X_train, y_train_scaled)
```

```python
y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)


print("training r2 score: " , r2_score(y_train_scaled, y_train_pred_scaled))
print("testing r2 score: " , r2_score(y_test_scaled, y_test_pred_scaled))
#overfitting
```

```python
kernels = ["linear", "poly", "rbf", "sigmoid"]

for k in kernels:
    model = SVR(kernel = k)
    model.fit(X_train, y_train_scaled)
    
    y_test_pred_scaled = model.predict(X_test)
    y_train_pred_scaled = model.predict(X_train)

    print(f"{k} \ntraining r2 score: " , r2_score(y_train_scaled, y_train_pred_scaled))
    print("testing r2 score: " , r2_score(y_test_scaled, y_test_pred_scaled))
    print("\n")
```

# \# hyperparameter tuning using GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "C" : [1,2, 5, 10, 50, 100],
    "kernel" : ["linear", "rbf", "poly", "sigmoid"],
    "epsilon" : [0.01, 0.1, 0.2, 0.3, 0.5]
}
```

```python
model = SVR()

grid_search = GridSearchCV(model, param_grid, cv = 5, scoring = "r2")

grid_search.fit(X_train, y_train_scaled)
```

```python
print("best params : ", grid_search.best_params_)
print("best score : ", grid_search.best_score_)
```

```python
best_model = SVR(kernel="linear", C = 10, epsilon=0.1)
best_model.fit(X_train, y_train_scaled)
```

```python
y_test_pred_scaled = best_model.predict(X_test)
y_train_pred_scaled = best_model.predict(X_train)


print("training r2 score: " , r2_score(y_train_scaled, y_train_pred_scaled))
print("testing r2 score: " , r2_score(y_test_scaled, y_test_pred_scaled))
```
