# \# Logistic Regression Scratched Implementation


```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def _sigmoid(self, z):
        return (1 / (1 + np.exp(-z)))

    def fit(self,X,y): 
        
        m,n =  X.shape 

        # step 1
        self.bias = 0
        self.weights = np.zeros(n) # array of zeros

        # Gradient Descent
        for i in range(self.n_iter) :

            z = self.bias + np.dot(X, self.weights)
            y_pred = self._sigmoid(z)
            
            db = (1/m) * y_pred-y # derivative with respect to bias
            dw = (1/m) * np.dot(X.T, (y_pred-y))# derivative with respect to coeficient

            # step 4
            self.bias -= self.lr * db
            self.weights -= self.lr * dw
               
    def get_probability(self,x):
        z = self.bias + np.dot(X, self.weights)
        return self._sigmoid(z)
        
    def predict(self,X, threshold=0.5):
        probabilities = self.get_probability(X)
        y_pred = probabilities >= threshold
        y_pred = y_pred.astype(int)
        return y_pred
```

```python
X = np.array([[1,2],[2,3],[3,4],[4,5]])
y = np.array([0,0,1,1])
```

```python
model = LogisticRegression()
model.fit(X,y)

y_pred = model.predict(X)
print(y_pred)
print(model.get_probability(X))
```
