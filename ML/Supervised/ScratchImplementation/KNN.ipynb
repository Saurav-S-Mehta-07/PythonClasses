# \# KNN Implementation from scratch

```python
import numpy as np
```

```python
# binary classification

class KNeighborsClassifier:
    def __init__(self, n_neighbors=3):
        self.k = n_neighbors

    def _euclidean_dist(self, x1, x2):
        return np.sqrt(np.sum((x1-x2)**2))
        

    def fit(self,X,y):
        self.X_train = X
        self.y_train = y

    def _predict_one(self, x):
        
        distances = [self._euclidean_dist(x, x_train) for x_train in self.X_train]
        
        k_neigh_indices = np.argsort(distances)[:self.k]

        knn_classes = [self.y_train[i] for i in k_neigh_indices]

        majority_class = np.argmax(np.bincount(knn_classes))

        return majority_class
        
       

    def predict(self,X):
        y_pred = [self._predict_one(x) for x in X]
        return np.array(y_pred)
        
    
```

```python
# data
X_train = np.array([
    [1,2],
    [2,3],
    [3,3],
    [6,5],
    [7,7],
])

y_train = np.array([0,0,0,1,1])

X_test = np.array([
    [2,2],
    [6,6]
])
```

```python
model = KNeighborsClassifier()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(y_pred)
```

# \# KNN Regression scratch Implementation

```python
import numpy as np

class KNeighborsRegressor:

    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        predictions = []

        for x in X:
            # Euclidean distance
            distances = np.sqrt(np.sum((self.X_train - x) ** 2, axis=1))

            # Indices of k nearest neighbors
            k_indices = np.argsort(distances)[:self.k]

            # Average target value
            prediction = np.mean(self.y_train[k_indices])

            predictions.append(prediction)

        return np.array(predictions)
```

```python
# data
X_train = np.array([
    [1,2],
    [2,3],
    [3,3],
    [6,5],
    [7,7],
])

y_train = np.array([1,1,2,3,3])

X_test = np.array([
    [2,2],
    [6,6]
])
```

```python
model = KNeighborsRegressor()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(y_pred)
```
